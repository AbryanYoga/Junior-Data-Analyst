```python
import pandas as pd
import numpy as np

df_raw = pd.read_csv('Persentase Penduduk Miskin (P0) Menurut Provinsi dan Daerah, 2025.csv', skiprows=4, header=None)

columns = [
    'Provinsi',
    'Perkotaan_Sem1', 'Perkotaan_Sem2', 'Perkotaan_Tahunan',
    'Perdesaan_Sem1', 'Perdesaan_Sem2', 'Perdesaan_Tahunan',
    'Jumlah_Sem1', 'Jumlah_Sem2', 'Jumlah_Tahunan'
]
df_raw.columns = columns

df = df_raw.dropna(subset=['Provinsi']).copy()
df = df[df['Provinsi'] != 'INDONESIA'].copy()

num_cols = ['Perkotaan_Sem1', 'Perkotaan_Sem2', 'Perdesaan_Sem1', 'Perdesaan_Sem2', 'Jumlah_Sem1', 'Jumlah_Sem2']
for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print("Jumlah provinsi unik (nunique):", df['Provinsi'].nunique())
print("Contoh 5 Provinsi (unique):", df['Provinsi'].unique()[:5])

stats_dict = {}
for col in num_cols:
    mode_val = df[col].mode()
    mode_res = mode_val[0] if not mode_val.empty else np.nan

    stats_dict[col] = {
        'Mean': df[col].mean(),
        'Median': df[col].median(),
        'Modus': mode_res,
        'Minimum': df[col].min(),
        'Maksimum': df[col].max(),
        'Standar Deviasi': df[col].std(),
        'Kuartil 1 (Q1)': df[col].quantile(0.25),
        'Kuartil 2 (Q2)': df[col].quantile(0.50),
        'Kuartil 3 (Q3)': df[col].quantile(0.75)
    }

summary_df = pd.DataFrame(stats_dict).round(2)

summary_df.to_excel('Statistical Summary.xlsx')
print("File 'Statistical Summary.xlsx' berhasil dibuat!")
```
