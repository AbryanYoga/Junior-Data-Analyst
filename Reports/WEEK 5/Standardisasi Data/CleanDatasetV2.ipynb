```python
import pandas as pd
import numpy as np

filename = 'Persentase Penduduk Miskin (P0) Menurut Provinsi dan Daerah, 2025.csv'
df_raw = pd.read_csv(filename, header=None)

print("=== PROSES STANDARDISASI DATA ===")

df_clean = df_raw.iloc[4:].copy().reset_index(drop=True)

new_columns = [
    'provinsi',
    'perkotaan_sem1_maret',
    'perkotaan_sem2_september',
    'perkotaan_tahunan',
    'perdesaan_sem1_maret',
    'perdesaan_sem2_september',
    'perdesaan_tahunan',
    'jumlah_sem1_maret',
    'jumlah_sem2_september',
    'jumlah_tahunan'
]
df_clean.columns = new_columns

df_clean['provinsi'] = df_clean['provinsi'].astype(str).str.strip().str.upper()

numeric_cols = [col for col in new_columns if col != 'provinsi']

for col in numeric_cols:
    df_clean[col] = df_clean[col].astype(str).str.strip()
    df_clean[col] = df_clean[col].replace('-', np.nan)
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

output_csv = 'Clean_Dataset_v2.csv'
df_clean.to_csv(output_csv, index=False)

print(f"\n✅ SUCCESS: Dataset berhasil distandardisasi dan disimpan sebagai '{output_csv}'!")
print("\n--- Pratinjau 5 Baris Pertama Dataset v2 ---")
print(df_clean.head())

print("\n--- Info Tipe Data & Missing Values ---")
print(df_clean.info())
```
