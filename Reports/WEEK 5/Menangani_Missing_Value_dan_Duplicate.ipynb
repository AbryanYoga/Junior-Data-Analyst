```python
import pandas as pd
from google.colab import files

#Load Dataset
df = pd.read_csv('retail_sales_dataset.csv')

#Audit Data Awal
print("=== INFORMASI DATASET AWAL ===")
print(f"Jumlah Baris : {df.shape[0]}")
print(f"Jumlah Kolom : {df.shape[1]}")
print(f"Total Duplikat: {df.duplicated().sum()}")
print("\n--- Missing Value Per Kolom ---")
print(df.isnull().sum())

#Handling Duplicate Data
df_clean = df.drop_duplicates()

#Handling Missing Values (Median untuk Numerik, Modus untuk Kategorikal)
for col in df_clean.columns:
    if df_clean[col].isnull().sum() > 0:
        if df_clean[col].dtype in ['float64', 'int64']:
            df_clean[col] = df_clean[col].fillna(df_clean[col].median())
        else:
            df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

#Cek Ulang Hasil Pembersihan
print("\n=== HASIL SETELAH CLEANING ===")
print(f"Jumlah Baris setelah Drop Duplicate: {df_clean.shape[0]}")
print("Sisa Missing Value:")
print(df_clean.isnull().sum())

#Simpan Output Dataset Clean v1
df_clean.to_csv('retail_sales_clean_v1.csv', index=False)
print("\n[BERHASIL] File 'retail_sales_clean_v1.csv' berhasil dibuat.")
```
